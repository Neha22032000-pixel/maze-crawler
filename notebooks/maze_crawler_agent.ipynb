# Maze Crawler Agent

This notebook documents the tested Kaggle submission for the Maze Crawler competition.

For this competition, the submission artifact is `main.py` with an `agent(obs, config)` function, not a CSV file.

## Local Test

Install the environment if needed:

```powershell
pip install kaggle-environments
```

Then run a local match against the random baseline.

````python
from kaggle_environments import make

env = make("crawl", configuration={"randomSeed": 42}, debug=True)
env.run(["../main.py", "random"])

[(i, state.reward, state.status) for i, state in enumerate(env.steps[-1])]
````

## Submit

Submit the root `main.py` file:

```powershell
kaggle competitions submit maze-crawler -f main.py -m "tested starter agent v1"
```

Check status:

```powershell
kaggle competitions submissions maze-crawler
```
